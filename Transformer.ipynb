```python
import torch
import torch.nn as nn
import math
```

```python
max_seq = 3
dim_embeddings = 128
W_size = 32

test_data = torch.tensor([
    [0, 1, 2],
    [1, 2, 0],
    [2, 0, 1]
], dtype=torch.long)
```

```python
torch.manual_seed(42)

class MultiHeadAttention(nn.Module):
    def __init__(self, max_seq, dim_embeddings, W_size, vocab_size):
        super().__init__()

        # Configuración del modelo
        self.max_seq = max_seq
        self.dim_embeddings = dim_embeddings
        self.W_size = W_size
        self.vocab_size = vocab_size

        # Parámetros entrenables
        self.Wq = nn.Parameter(torch.rand(dim_embeddings, W_size))
        self.Wk = nn.Parameter(torch.rand(dim_embeddings, W_size))
        self.Wv = nn.Parameter(torch.rand(dim_embeddings, W_size))
        self.linear = nn.Linear(self.W_size, self.dim_embeddings)

        # Embeddings
        self.embeddings = nn.Embedding(vocab_size, dim_embeddings)
        self.pos_embeddings = nn.Embedding(max_seq, dim_embeddings)

    def forward(self, data):
        emb = self.embeddings(data) + self.pos_embeddings(torch.arange(self.max_seq))

        Q = emb @ self.Wq
        K = emb @ self.Wk
        V = emb @ self.Wv

        output_matrix = torch.softmax((Q @ torch.transpose(K, -1, -2) / (self.W_size) ** 0.5), dim=2) @ V
        print((Q @ torch.transpose(K, -1, -2) / (self.W_size) ** 0.5).shape)      

        return self.linear(output_matrix) + emb
```

```python
test = MultiHeadAttention(max_seq, dim_embeddings, W_size, 100)
test.forward(test_data)
```
